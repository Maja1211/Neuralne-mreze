```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
import os
import warnings
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score, precision_score, recall_score, f1_score
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Conv2D, MaxPooling2D, Flatten, Dense, Dropout, Rescaling
from tensorflow.keras.preprocessing.image import load_img
from IPython.display import display, Markdown
from collections import Counter

warnings.filterwarnings('ignore')
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '3'

dataset_dir = "Vehicles"
batch_size = 32
img_size = (128, 128)
seed = 42

data_augmentation = tf.keras.Sequential([
    tf.keras.layers.RandomFlip("horizontal"),
    tf.keras.layers.RandomRotation(0.05),
    tf.keras.layers.RandomZoom(0.05),
])

full_dataset = tf.keras.preprocessing.image_dataset_from_directory(
    dataset_dir,
    image_size=img_size,
    batch_size=1,
    shuffle=True,
    seed=seed
)
class_names = full_dataset.class_names

images, labels = [], []
for img, label in full_dataset:
    images.append(img[0])
    labels.append(label[0])
images = tf.stack(images)
labels = tf.stack(labels)

total_size = len(images)
val_size = test_size = total_size // 10
train_size = total_size - val_size - test_size
print(f"Broj uzoraka:\n- Trening skup: {train_size}\n- Validacioni skup: {val_size}\n- Test skup: {test_size}")

train_imgs = images[:train_size]
train_lbls = labels[:train_size]
val_imgs = images[train_size:train_size + val_size]
val_lbls = labels[train_size:train_size + val_size]
test_imgs = images[train_size + val_size:]
test_lbls = labels[train_size + val_size:]

train_ds = tf.data.Dataset.from_tensor_slices((train_imgs, train_lbls))
val_ds = tf.data.Dataset.from_tensor_slices((val_imgs, val_lbls))
test_ds = tf.data.Dataset.from_tensor_slices((test_imgs, test_lbls))

normalization_layer = Rescaling(1./255)
train_ds = train_ds.map(lambda x, y: (data_augmentation(normalization_layer(x)), y))
val_ds = val_ds.map(lambda x, y: (normalization_layer(x), y))
test_ds = test_ds.map(lambda x, y: (normalization_layer(x), y))

AUTOTUNE = tf.data.AUTOTUNE
train_ds = train_ds.shuffle(1000).batch(batch_size).cache().prefetch(buffer_size=AUTOTUNE)
val_ds = val_ds.batch(batch_size).cache().prefetch(buffer_size=AUTOTUNE)
test_ds = test_ds.batch(batch_size).cache().prefetch(buffer_size=AUTOTUNE)

model = Sequential([
    Input(shape=(128, 128, 3)),
    Conv2D(32, (3, 3), activation='relu'),
    MaxPooling2D(),
    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D(),
    Conv2D(128, (3, 3), activation='relu'),
    MaxPooling2D(),
    Flatten(),
    Dropout(0.5),
    Dense(128, activation='relu'),
    Dense(len(class_names), activation='softmax')
])
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

history = model.fit(train_ds, validation_data=val_ds, epochs=15)

y_true, y_pred = [], []
for images, labels in test_ds:
    preds = model.predict(images)
    y_true.extend(labels.numpy())
    y_pred.extend(np.argmax(preds, axis=1))

print(classification_report(y_true, y_pred, target_names=class_names))
acc = accuracy_score(y_true, y_pred)
prec = precision_score(y_true, y_pred, average='weighted', zero_division=0)
rec = recall_score(y_true, y_pred, average='weighted', zero_division=0)
f1 = f1_score(y_true, y_pred, average='weighted', zero_division=0)

display(Markdown(f"""
### Rezultati metrika (test skup)

- Tačnost (Accuracy): {acc:.4f}  
- Preciznost (Precision): {prec:.4f}  
- Osetljivost (Recall): {rec:.4f}  
- F1 skor: {f1:.4f}
"""))

cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
            xticklabels=class_names, yticklabels=class_names)
plt.xlabel("Predviđeno")
plt.ylabel("Stvarno")
plt.title("Matrica konfuzije - Test skup")
plt.tight_layout()
plt.show()

y_true_train, y_pred_train = [], []
for images, labels in train_ds:
    preds = model.predict(images)
    y_true_train.extend(labels.numpy())
    y_pred_train.extend(np.argmax(preds, axis=1))

cm_train = confusion_matrix(y_true_train, y_pred_train)
plt.figure(figsize=(8, 6))
sns.heatmap(cm_train, annot=True, fmt="d", cmap="Greens",
            xticklabels=class_names, yticklabels=class_names)
plt.xlabel("Predviđeno")
plt.ylabel("Stvarno")
plt.title("Matrica konfuzije - Trening skup")
plt.tight_layout()
plt.show()

correct, incorrect = [], []
for images, labels in test_ds:
    preds = model.predict(images)
    pred_labels = np.argmax(preds, axis=1)
    for i in range(len(labels)):
        if len(correct) < 3 and pred_labels[i] == labels[i]:
            correct.append((images[i], class_names[pred_labels[i]]))
        elif len(incorrect) < 3 and pred_labels[i] != labels[i]:
            incorrect.append((images[i], class_names[pred_labels[i]], class_names[labels[i]]))
        if len(correct) >= 3 and len(incorrect) >= 3:
            break
    if len(correct) >= 3 and len(incorrect) >= 3:
        break

plt.figure(figsize=(12, 4))
for i, (img, label) in enumerate(correct):
    plt.subplot(1, 3, i + 1)
    img_np = img.numpy()
    img_np = (img_np * 255).astype("uint8") if img_np.max() <= 1.0 else img_np.astype("uint8")
    plt.imshow(img_np)
    plt.title(f"Tačno: {label}")
    plt.axis("off")
plt.suptitle("Dobro klasifikovane slike")
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 4))
for i, (img, pred, true) in enumerate(incorrect):
    plt.subplot(1, 3, i + 1)
    img_np = img.numpy()
    img_np = (img_np * 255).astype("uint8") if img_np.max() <= 1.0 else img_np.astype("uint8")
    plt.imshow(img_np)
    plt.title(f"Pogrešno: {pred}\nTačno: {true}")
    plt.axis("off")
plt.suptitle("Loše klasifikovane slike")
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 5))
plt.plot(history.history['accuracy'], label='Trening tačnost')
plt.plot(history.history['val_accuracy'], label='Validaciona tačnost')
plt.title('Tačnost tokom treniranja')
plt.xlabel('Epohe')
plt.ylabel('Tačnost')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 5))
plt.plot(history.history['loss'], label='Trening gubitak')
plt.plot(history.history['val_loss'], label='Validacioni gubitak')
plt.title('Gubitak tokom treniranja')
plt.xlabel('Epohe')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

all_labels = train_lbls.numpy()
label_counts = Counter(all_labels)
label_names = [class_names[i] for i in label_counts.keys()]
counts = list(label_counts.values())

plt.figure(figsize=(8, 5))
sns.barplot(x=label_names, y=counts, palette='viridis')
plt.title("Broj uzoraka po klasama (trening skup)")
plt.xlabel("Klase")
plt.ylabel("Broj uzoraka")
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 8))
for i, class_name in enumerate(class_names):
    class_path = os.path.join(dataset_dir, class_name)
    img_name = sorted(os.listdir(class_path))[0]
    img_path = os.path.join(class_path, img_name)
    img = load_img(img_path, target_size=img_size)
    plt.subplot(2, 3, i + 1)
    plt.imshow(img)
    plt.title(f"{class_name}")
    plt.axis("off")
plt.suptitle("Primer po jedna slika iz svake klase")
plt.tight_layout()
plt.show()

model.summary()
```
